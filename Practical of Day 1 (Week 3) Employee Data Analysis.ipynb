```python
# Average employee age

average_age = df["Age"].mean()
print(f"The Average employee age is {average_age:.1f}")


plt.bar(df["Age"], df["Age"])
plt.axhline(average_age, linestyle="--", label=f"Average Age: {average_age:.1f}")

plt.xlabel("Employee")
plt.ylabel("Age")
plt.title("Employee Age and Average Age")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

# Department Analysis

```python
# Number of employees in each department

df.head()
```

```python
# Number of employees in each department

number_employee_department = df.groupby("Department")["Employee_ID"].nunique().sort_values()
print(number_employee_department)


number_employee_department.plot(kind="pie",figsize=(8, 5),autopct="%1.1f%%")

plt.ylabel("")  
plt.title("Number of Employees by Department")
plt.show()
```

```python
# Average salary by department

average_salary = df.groupby("Department")["Salary"].mean().sort_values().round()
print(average_salary)


plt.scatter(
    average_salary.index,
    average_salary.values
)

plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.title("Average Salary by Department")
plt.xticks(rotation=45)
plt.show()
```

```python
# Average age by department

average_age = df.groupby("Department")["Age"].mean().sort_values().round()
print(average_age)


plt.figure(figsize=(9, 5.5))

bars = plt.bar(average_age.index, average_age.values, color='Black', width=0.6)

for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2.0, 
        yval + 0.5,                    
        f'{int(yval)} yrs', 
        ha='center', 
        va='bottom', 
        fontsize=9, 
        weight='bold',
        color = "Red"
    )

plt.xlabel("Department", fontsize=11, labelpad=10)
plt.ylabel("Average Age (Years)", fontsize=11, labelpad=10)
plt.title("Average Age by Department", fontsize=14, pad=15, weight='bold')

plt.xticks(rotation=45, ha='right')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# Which department has the highest number of employees

top_department = df.groupby('Department')['Employee_ID'].nunique().idxmax()
print(f"The department with the highest number of employees is: {top_department}")


sns.countplot(data=df, x='Department')

plt.title('Number of Employees by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.show()
```

```python
# Which employees have unusually high/low salaries?

Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

salary = df[(df["Salary"] < lower_bound) | (df["Salary"] > upper_bound)]
print(salary)
```
