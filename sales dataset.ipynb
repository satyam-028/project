# Exploratory Data Analysis — EDA

```python
# Which category has the highest sales?

Category_sales = df.groupby('Category')["Sales"].sum()

highest_category = Category_sales.idxmax()
highest_value = Category_sales.max()

print(f"The category with the highest sales is '{highest_category}' with a total of {highest_value:.2f}")
```

```python
df.head(2)
```

```python
plt.figure(figsize=(10, 6))

Category_sales.plot(kind='bar')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Which category has the lowest sales?

category_sales = df.groupby('Category')["Sales"].sum()

low_category = category_sales.idxmin()
low_value = category_sales.min()

print(f"The category with the highest sales is '{low_category}' with a total of {low_value:.2f}")
```

```python
plt.figure(figsize=(10, 6))

Category_sales.plot(kind='pie', autopct='%1.1f%%')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Which region performs best?

region_sales = df.groupby('Region')['Sales'].sum()

best_region = region_sales.idxmax()
best_value = region_sales.max()

print(
    f"The best performing region is '{best_region}' with a total of {best_value:.2f}")
```

```python
# Which region performs worst?

regions_sales = df.groupby('Region')['Sales'].sum()

worst_region = regions_sales.idxmax()
worst_value = regions_sales.max()

print(
    f"The best performing region is '{worst_region}' with a total of {worst_value:.2f}")
```

```python
plt.figure(figsize=(10, 6))

regions_sales.plot(kind='pie',autopct='%1.1f%%')

plt.title('Total Sales by Region')
plt.ylabel('') 
plt.tight_layout()
plt.show()
```

# Profit Analysis

```python
# Total profit    # Average profit



total_profit = df['Profit'].sum()
average_profit = df['Profit'].mean()

print(f"Total Business Profit: ${total_profit:,.2f}")
print(f"Average Profit per Transaction: ${average_profit:,.2f}\n")
```

```python
# Most profitable products   # Least profitable products

product_profit_totals = df.groupby('Product_Name')['Profit'].sum()

most_profitable = product_profit_totals.sort_values(ascending=False).head(5)
least_profitable = product_profit_totals.sort_values(ascending=True).head(5)

print("--- Top 5 Most Profitable Products ---")
print(most_profitable.to_string(), "\n")

print("--- Bottom 5 Least Profitable Products ---")
print(least_profitable.to_string(), "\n")
```
